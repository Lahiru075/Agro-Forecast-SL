```python
import pandas as pd
from pathlib import Path

raw_path = Path("../data/row/researchData.xlsx")
df = pd.read_excel(raw_path)

print(f"Dataset Loaded Successfully! Total Rows & Columns: {df.shape}")
print("--- Raw Dataset Head ---")
display(df.head())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Yala vs Maha seasonal production comparison boxplot
plt.figure(figsize=(10, 6))
sns.boxplot(x='Season', y='Production', data=df)
plt.title("Yala vs Maha Seasonal Production Comparison")
plt.yscale('log')
plt.show()
```
